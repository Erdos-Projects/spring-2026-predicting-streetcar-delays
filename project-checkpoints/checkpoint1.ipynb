# Checkpoint 1: Problem Definition, Data Gathering, and KPIs

## Problem Definition
Our goal in this project is to create a predictor that determines the delay time of a streetcar from the TTC (Toronto Transit Commision) given certain features which necessarily will include the station name, date, and time. Due to the forthcoming Fifa world cup coming to Toronto, the amount of traffic will be large, and therefore the purpose of the predictor will be to direct policy for the city of Toronto and the TTC. 

Additional details:
- Since the predictor is for directing public policy, it is a long-term predictor (rather than a real-time predictor). 
- In particular, this predictor will *not* be expected to perform well on short term (within a day) predictions. 
- Our target comes from the TTC Streetcar delay data. This is a csv file containing delay incidents from the TTC for the years 2014-2025, and includes the amount of minutes of delay caused by an incident.
- There is also a real-time TTC data API for more recent information.

### Data Assessment
- The streetcar delay data is from 2014--2025, and has few missing entries, meaning that it is large enough to account for seasonal trends in delays. 
- The delay data seems to be 'hand recorded' so it may not be 100% precise, and has some typos that need to be fixed. 


### Ethical and legal considerations
Other than licensing concerns, our data does not appear to have any ethical concerns. In particular, there is no sensitive data in our data sets.
